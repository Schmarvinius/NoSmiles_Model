```python
# Smile and Laughter Detection: Methodology Implementation
# -------------------------------------------------------
# This notebook implements the methodology for real-time smile and laughter detection
# using the FER-2013 dataset and a pre-trained ResNet18 model with progressive unfreezing.

# 1. Imports and Setup
import os
import random
import numpy as np
import torch
import torch.nn as nn
import torch.optim as optim
from torch.optim.lr_scheduler import ReduceLROnPlateau
from torch.utils.data import DataLoader, Subset, ConcatDataset
from torchvision import datasets, transforms, models
from sklearn.model_selection import StratifiedKFold
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    confusion_matrix,
)
import matplotlib.pyplot as plt
import seaborn as sns
from tqdm import tqdm
import time

print("Success")
```

```python
# Reproducibility
SEED = 42
torch.manual_seed(SEED)
np.random.seed(SEED)
random.seed(SEED)
torch.backends.cudnn.deterministic = True
torch.backends.cudnn.benchmark = False
```

```python
# 2. Data Collection and Preparation
# ----------------------------------
# Download FER-2013 from Kaggle and place it in './data/fer2013'
# Only 'happy' (smile/laugh) and 'neutral' (no laugh) classes are used.

# Helper: Map FER-2013 labels to binary (0: No Laugh, 1: Laugh)
FER_LABELS = {
    0: "Angry",
    1: "Disgust",
    2: "Fear",
    3: "Happy",    # Laugh/Smile
    4: "Sad",
    5: "Surprise",
    6: "Neutral",  # No Laugh
}

# Only keep 'Happy' and 'Neutral'
TARGET_LABELS = [3, 6]
LABEL_MAP = {3: 1, 6: 0}  # 1: Laugh/Smile, 0: No Laugh

# Custom Dataset to filter and relabel FER-2013
class FER2013Binary(datasets.FakeData):
    def __init__(self, root, split, transform=None):
        super().__init__()
        self.data = []
        self.targets = []
        # Load FER-2013 CSV
        import pandas as pd

        csv_path = os.path.join(root, "fer2013.csv")
        df = pd.read_csv(csv_path)
        df = df[df["Usage"] == split]
        for _, row in df.iterrows():
            label = int(row["emotion"])
            if label in TARGET_LABELS:
                img = np.fromstring(row["pixels"], sep=" ", dtype=np.uint8).reshape(48, 48)
                self.data.append(img)
                self.targets.append(LABEL_MAP[label])
        self.transform = transform

    def __len__(self):
        return len(self.data)

    def __getitem__(self, idx):
        img = self.data[idx]
        label = self.targets[idx]
        img = np.stack([img] * 3, axis=-1)  # Convert to 3-channel
        if self.transform:
            img = self.transform(img)
        return img, label

# Data transforms: normalization, augmentation
train_transform = transforms.Compose(
    [
        transforms.ToPILImage(),
        transforms.RandomHorizontalFlip(),
        transforms.RandomRotation(10),
        transforms.RandomResizedCrop(48, scale=(0.8, 1.0)),
        transforms.ToTensor(),
        transforms.Normalize([0.5], [0.5]),
    ]
)
test_transform = transforms.Compose(
    [
        transforms.ToPILImage(),
        transforms.ToTensor(),
        transforms.Normalize([0.5], [0.5]),
    ]
)

# Load datasets
data_root = ""
train_dataset = FER2013Binary(data_root, "Training", transform=train_transform)
val_dataset = FER2013Binary(data_root, "PublicTest", transform=test_transform)
test_dataset = FER2013Binary(data_root, "PrivateTest", transform=test_transform)

# Undersample to balance classes
def undersample(dataset):
    from collections import Counter

    targets = np.array([label for _, label in dataset])
    class_counts = Counter(targets)
    min_count = min(class_counts.values())
    indices = []
    for c in class_counts:
        c_indices = np.where(targets == c)[0]
        indices.extend(np.random.choice(c_indices, min_count, replace=False))
    return Subset(dataset, indices)

train_dataset = undersample(train_dataset)
val_dataset = undersample(val_dataset)
test_dataset = undersample(test_dataset)
```

```python
# 3. Model Architecture and Fine-Tuning
# -------------------------------------
# Use pre-trained ResNet18, replace head for binary classification

def get_model():
    model = models.resnet18(pretrained=True)
    for param in model.parameters():
        param.requires_grad = False  # Freeze all layers
    num_ftrs = model.fc.in_features
    model.fc = nn.Sequential(
        nn.Dropout(0.5),
        nn.Linear(num_ftrs, 1),
    )
    return model
```

```python
# 4. Training Utilities
# ---------------------
def train_one_epoch(model, loader, criterion, optimizer, device):
    model.train()
    running_loss = 0.0
    preds, labels = [], []
    for x, y in loader:
        x, y = x.to(device), y.float().to(device)
        optimizer.zero_grad()
        out = model(x).squeeze(1)
        loss = criterion(out, y)
        loss.backward()
        optimizer.step()
        running_loss += loss.item() * x.size(0)
        preds.extend(torch.sigmoid(out).cpu().detach().numpy() > 0.5)
        labels.extend(y.cpu().numpy())
    acc = accuracy_score(labels, preds)
    return running_loss / len(loader.dataset), acc

def eval_model(model, loader, criterion, device):
    model.eval()
    running_loss = 0.0
    preds, labels = [], []
    with torch.no_grad():
        for x, y in loader:
            x, y = x.to(device), y.float().to(device)
            out = model(x).squeeze(1)
            loss = criterion(out, y)
            running_loss += loss.item() * x.size(0)
            probs = torch.sigmoid(out).cpu().numpy()
            preds.extend(probs > 0.5)
            labels.extend(y.cpu().numpy())
    acc = accuracy_score(labels, preds)
    prec = precision_score(labels, preds)
    rec = recall_score(labels, preds)
    f1 = f1_score(labels, preds)
    try:
        auc = roc_auc_score(labels, preds)
    except:
        auc = 0.0
    return running_loss / len(loader.dataset), acc, prec, rec, f1, auc, preds, labels
```

```python
# 5. Stratified K-Fold Cross-Validation
# -------------------------------------
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
BATCH_SIZE = 16
EPOCHS = 10
PATIENCE = 5

from torch.utils.data import Dataset

class SimpleDataset(Dataset):
    def __init__(self, dataset):
        self.dataset = dataset
        self.targets = [label for _, label in dataset]

    def __len__(self):
        return len(self.dataset)

    def __getitem__(self, idx):
        return self.dataset[idx]

full_dataset = SimpleDataset(train_dataset)
targets = np.array(full_dataset.targets)
skf = StratifiedKFold(n_splits=5, shuffle=True, random_state=SEED)

fold_metrics = []

for fold, (train_idx, val_idx) in enumerate(skf.split(np.zeros(len(targets)), targets)):
    print(f"\nFold {fold+1}/5")
    train_subset = Subset(full_dataset, train_idx)
    val_subset = Subset(full_dataset, val_idx)
    train_loader = DataLoader(train_subset, batch_size=BATCH_SIZE, shuffle=True)
    val_loader = DataLoader(val_subset, batch_size=BATCH_SIZE, shuffle=False)

    model = get_model().to(device)
    criterion = nn.BCEWithLogitsLoss()
    optimizer = optim.AdamW(
        [
            {"params": model.fc.parameters(), "lr": 1e-3},
            {"params": [p for n, p in model.named_parameters() if "fc" not in n], "lr": 1e-5},
        ],
        weight_decay=1e-2,
    )
    scheduler = ReduceLROnPlateau(optimizer, "max", patience=2, factor=0.5)
    best_f1 = 0
    patience_counter = 0

    for epoch in range(EPOCHS):
        train_loss, train_acc = train_one_epoch(
            model, train_loader, criterion, optimizer, device
        )
        val_loss, val_acc, val_prec, val_rec, val_f1, val_auc, _, _ = eval_model(
            model, val_loader, criterion, device
        )
        print(
            f"Epoch {epoch+1}: "
            f"Train Loss={train_loss:.4f}, Acc={train_acc:.4f} | "
            f"Val Loss={val_loss:.4f}, Acc={val_acc:.4f}, F1={val_f1:.4f}"
        )
        scheduler.step(val_f1)
        if val_f1 > best_f1:
            best_f1 = val_f1
            patience_counter = 0
            best_model_state = model.state_dict()
        else:
            patience_counter += 1
        if patience_counter >= PATIENCE:
            print("Early stopping.")
            break

    # Evaluate on validation fold
    model.load_state_dict(best_model_state)
    val_loss, val_acc, val_prec, val_rec, val_f1, val_auc, preds, labels = eval_model(
        model, val_loader, criterion, device
    )
    cm = confusion_matrix(labels, preds)
    fold_metrics.append(
        {
            "acc": val_acc,
            "prec": val_prec,
            "rec": val_rec,
            "f1": val_f1,
            "auc": val_auc,
            "cm": cm,
        }
    )
```

```python
# 6. Aggregate and Display Results
# --------------------------------
def print_metrics(metrics):
    print(
        f"Accuracy: {np.mean([m['acc'] for m in metrics]):.4f} ± {np.std([m['acc'] for m in metrics]):.4f}"
    )
    print(
        f"Precision: {np.mean([m['prec'] for m in metrics]):.4f} ± {np.std([m['prec'] for m in metrics]):.4f}"
    )
    print(
        f"Recall: {np.mean([m['rec'] for m in metrics]):.4f} ± {np.std([m['rec'] for m in metrics]):.4f}"
    )
    print(
        f"F1 Score: {np.mean([m['f1'] for m in metrics]):.4f} ± {np.std([m['f1'] for m in metrics]):.4f}"
    )
    print(
        f"AUC: {np.mean([m['auc'] for m in metrics]):.4f} ± {np.std([m['auc'] for m in metrics]):.4f}"
    )

print_metrics(fold_metrics)

# Plot confusion matrix for last fold
plt.figure(figsize=(4, 4))
sns.heatmap(fold_metrics[-1]["cm"], annot=True, fmt="d", cmap="Blues")
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.title("Confusion Matrix (Last Fold)")
plt.show()
```

```python
# 7. Final Training and Test Evaluation
# -------------------------------------
# Train on full train+val, test on test set

full_train_dataset = ConcatDataset([train_dataset, val_dataset])
full_train_loader = DataLoader(full_train_dataset, batch_size=BATCH_SIZE, shuffle=True)
test_loader = DataLoader(test_dataset, batch_size=BATCH_SIZE, shuffle=False)

model = get_model().to(device)
criterion = nn.BCEWithLogitsLoss()
optimizer = optim.AdamW(
    [
        {"params": model.fc.parameters(), "lr": 1e-3},
        {"params": [p for n, p in model.named_parameters() if "fc" not in n], "lr": 1e-5},
    ],
    weight_decay=1e-2,
)
scheduler = ReduceLROnPlateau(optimizer, "max", patience=2, factor=0.5)
best_f1 = 0
patience_counter = 0

for epoch in range(EPOCHS):
    train_loss, train_acc = train_one_epoch(
        model, full_train_loader, criterion, optimizer, device
    )
    val_loss, val_acc, val_prec, val_rec, val_f1, val_auc, _, _ = eval_model(
        model, test_loader, criterion, device
    )
    print(
        f"Epoch {epoch+1}: "
        f"Train Loss={train_loss:.4f}, Acc={train_acc:.4f} | "
        f"Test Loss={val_loss:.4f}, Acc={val_acc:.4f}, F1={val_f1:.4f}"
    )
    scheduler.step(val_f1)
    if val_f1 > best_f1:
        best_f1 = val_f1
        patience_counter = 0
        best_model_state = model.state_dict()
    else:
        patience_counter += 1
    if patience_counter >= PATIENCE:
        print("Early stopping.")
        break

model.load_state_dict(best_model_state)
test_loss, test_acc, test_prec, test_rec, test_f1, test_auc, preds, labels = eval_model(
    model, test_loader, criterion, device
)
print("\nFinal Test Metrics:")
print(f"Accuracy: {test_acc:.4f}")
print(f"Precision: {test_prec:.4f}")
print(f"Recall: {test_rec:.4f}")
print(f"F1 Score: {test_f1:.4f}")
print(f"AUC: {test_auc:.4f}")
```

```python
# 8. Inference Time Measurement
# -----------------------------
import time

model.eval()
n_samples = 100
sample_loader = DataLoader(test_dataset, batch_size=1, shuffle=True)
start = time.time()
with torch.no_grad():
    for i, (x, _) in enumerate(sample_loader):
        if i >= n_samples:
            break
        x = x.to(device)
        _ = model(x)
end = time.time()
avg_inference_time = (end - start) / n_samples
print(f"Average inference time per image: {avg_inference_time*1000:.2f} ms")
```
